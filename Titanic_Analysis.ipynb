```python
import pandas as pd
import numpy as np
df = pd.read_csv("Titanic_Dataset.csv")
```

```python
df_backup = df.copy()
print(df.head())
```

```python
print(df.info())
print(df.columns)
```

```python
df.dropna(inplace=True) # Delete Rows Have Nun Values
```

```python
print(df["Sex"].unique())
df['Sex'] = df['Sex'].replace ({"female":0,"male":1})
```

```python
print(df["Embarked"].unique())
df['Embarked'] = df['Embarked'].replace({'Q':1,'S':2,'C':3})
```

```python
df.drop(columns=['Name','Ticket','Cabin'], inplace=True) # Delete all rows whose type is not number 
```

```python
print(df.dtypes)
```

```python
X = df.drop("Survived",axis=1) # featuers : المميزات
y = df["Survived"] # Target : الهدف
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2,random_state=100
)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

knn = KNeighborsClassifier(n_neighbors=5) # انشاء النموذج *اقرب 5 اشخاص
knn.fit(X_train, y_train)
y_pred = knn.predict(X_test) # التنبؤ على بيانات الاختبار
accuracy = accuracy_score(y_test, y_pred) # حساب الدقة

print("Accuracy:", accuracy)
```
